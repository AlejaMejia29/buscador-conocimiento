Este cuaderno **no reemplaza la aplicación**: solo la descarga del repositorio y la abre con un enlace temporal de Gradio, para probarla sin instalar Python en el computador. La aplicación es la carpeta `app/`.

**Antes de ejecutar**, en *Secretos* (icono de llave) deben existir y tener acceso:

| Secreto | Valor |
|---|---|
| `GROQ_API_KEY` | Clave de Groq |
| `SUPABASE_URL` | Project URL de Supabase |
| `SUPABASE_PUBLISHABLE_KEY` | Clave publicable (`sb_publishable_...`) |
| `CORPUS_PROPIETARIO_ID` | UUID de la cuenta que cargó el corpus (ver README, «Preparar el corpus») |

El correo y la contraseña **no** van en los secretos: cada persona los escribe en la pantalla de ingreso de la aplicación.

```python
#@title 1. Descargar el repositorio e instalar dependencias (2 a 4 minutos)
!rm -rf buscador-conocimiento
!git clone -q https://github.com/AlejaMejia29/buscador-conocimiento
!pip -q install -r buscador-conocimiento/app/requirements.txt
print("Listo.")
```

```python
#@title 2. Configuración desde los secretos de Colab
import os
from google.colab import userdata

for nombre in ["GROQ_API_KEY", "SUPABASE_URL", "SUPABASE_PUBLISHABLE_KEY", "CORPUS_PROPIETARIO_ID"]:
    os.environ[nombre] = userdata.get(nombre)
CORPUS_ID = "buscador_conocimiento-29dccc23ac01" #@param {type:"string"}
os.environ["CORPUS_ID"] = CORPUS_ID
os.environ["EMBEDDINGS_REVISION"] = "e8f8c211226b894fcb81acc59f3b34ba3efd5f42"
os.environ["GROQ_MODEL"] = "openai/gpt-oss-20b"
os.environ["GRADIO_SERVER_NAME"] = "0.0.0.0"
os.environ["GRADIO_SHARE"] = "1"      # enlace temporal *.gradio.live
os.environ["MODO_PRUEBAS"] = "1"      # muestra la casilla para simular la caída del MCP
print("Configuración lista. Corpus:", CORPUS_ID)
```
